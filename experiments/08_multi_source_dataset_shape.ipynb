# All datasets: shape before clustering

Measures each dataset's shape with `sc4020.datasets.describe_dataset.describe_shape`, to decide which
validity metrics can be trusted on it (report Section 3.2, Table `tab:metric_choice`). Labels, where they
exist, are used only to describe the classes, never to tune a method.

```python
import warnings

import pandas as pd

from sc4020.datasets import load_dataset
from sc4020.datasets.describe_dataset import describe_shape
from sc4020.paths import RESULTS_DIR

warnings.filterwarnings("ignore")

TOPIC = "datasets"
REPORT_ID_BY_DATASET = {
    "make_moons": "D1",
    "earthquakes_2023": "D2",
    "uber_pickups": "D3",
    "ecoli": "D4",
    "fashion_mnist": "D5",
    "wine": "D6",
    "hdb_resale": "D8",
}
```

## Shape of every dataset

`density_spread` is the 90th over the 10th percentile of the distance to the 10th neighbour (1 means even
density); `far_outlier_share` the points whose distance lies beyond Q3 + 3 IQR. With labels:
`label_silhouette` and `label_davies_bouldin` score the classes as a clustering, and a 1-nearest-neighbour
accuracy well above the nearest-centroid accuracy means the classes are not convex.

```python
rows = []
for dataset_name, report_id in REPORT_ID_BY_DATASET.items():
    ds = load_dataset(dataset_name)
    rows.append({"report_id": report_id, "dataset": dataset_name, **describe_shape(ds.data, ds.target)})
shapes = pd.DataFrame(rows)
output_path = RESULTS_DIR / TOPIC / "dataset_shape.csv"
shapes.to_csv(output_path, index=False)
shapes.round(3).T
```
